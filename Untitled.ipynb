```python
from datetime import date
from django.contrib.auth.models import User
from listings.models import Listing, Category

# 1. Пользователь и категория (без дублей)
user, _ = User.objects.get_or_create(username='test_user', defaults={'email': 'test@dev.com'})
cat, _ = Category.objects.get_or_create(name='Платья')

# 2. Создаём объявление с датами
listing = Listing.objects.create(
    owner=user,
    title='Летнее платье',
    description='Отличное состояние, размер M',
    price=1500.00,
    location='Москва',
    contact_info='tg: @test',
    rental_start=date(2026, 9, 15),  # ← обязательно
    rental_end=date(2026, 9, 25),    # ← обязательно
    status='pending',
    is_moderated=False
)
listing.categories.add(cat)
print(f"✅ Создано: {listing} | ID: {listing.pk}")
```

```python
from django.db.models import Q

# 1. Только одобренные (как в публичном списке)
qs_public = Listing.objects.filter(is_moderated=True, status='approved')
print(f"🟢 Опубликовано: {qs_public.count()}")

# 2. На модерации (как видит автор)
qs_pending = Listing.objects.filter(owner=user, status='pending')
print(f"⏳ На модерации: {qs_pending.count()}")

# 3. Смотрим, какой SQL Django отправит в БД
print("\n🔍 Скомпилированный SQL:")
print(qs_public.query)
```
